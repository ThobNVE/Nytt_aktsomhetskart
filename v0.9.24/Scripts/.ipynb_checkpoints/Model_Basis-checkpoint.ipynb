```python
## GETTING NBF BOUNDARY FROM FKB DATA

def get_nbf_bounds(fkb_nedborfelt: gpd.GeoDataFrame, vassdrag_nr: str) -> tuple:
    """
    Get the exact bounding box coordinates for a specific drainage basin (nedbørfelt)
    based on its vassdragNr and return the basin polygon.
    
    Args:
        fkb_nedborfelt: GeoDataFrame containing watershed data
        vassdrag_nr: The vassdragNr identifier for the desired drainage basin
        
    Returns:
        tuple: ((xmin, ymin, xmax, ymax), basin_gdf) - bounds and the basin GeoDataFrame
    """
    # Filter the dataset to get only the specified drainage basin
    basin = fkb_nedborfelt[fkb_nedborfelt['vassdragNr'] == vassdrag_nr]
    
    if basin.empty:
        raise ValueError(f"No drainage basin found with vassdragNr: {vassdrag_nr}")
    
    # Get the basin name
    basin_name = basin.iloc[0]['navnNedbf'] if 'navnNedbf' in basin.columns else "Unknown"
    
    # Get the exact bounds of the filtered basin
    xmin, ymin, xmax, ymax = basin.total_bounds
    
    print(f"Bounds for vassdragNr {vassdrag_nr} ({basin_name}):")
    print(f"xmin: {xmin}, ymin: {ymin}, xmax: {xmax}, ymax: {ymax}")
    
    return (xmin, ymin, xmax, ymax), basin
```

```python
def modify_dtm_at_rivers(
    dtm_path: str, 
    river_lines: gpd.GeoDataFrame, 
    basin_boundary: gpd.GeoDataFrame = None,
    river_depth: float = 5.0,
    output_path: str = None
) -> tuple:
    """
    Subtract a specified depth from DTM cells where rivers pass through.
    
    Args:
        dtm_path: Path to the Digital Terrain Model raster file
        river_lines: GeoDataFrame containing river line geometries
        basin_boundary: Optional GeoDataFrame with polygon defining processing extent
        river_depth: Depth in meters to subtract from river cells (default: 5.0)
        output_path: Optional path to save the modified DTM (default: None)
        
    Returns:
        tuple: (modified_dtm, transform, crs) - The modified DTM array, its transform and CRS
    """
    # Open the DTM raster
    with rasterio.open(dtm_path) as src:
        # Read the data and get metadata
        dtm = src.read(1)
        transform = src.transform
        crs = src.crs
        profile = src.profile.copy()
        
        # Create a mask for the basin boundary if provided
        if basin_boundary is not None:
            # Reproject basin boundary to match DTM CRS if needed
            if basin_boundary.crs != crs:
                basin_boundary = basin_boundary.to_crs(crs)
            
            # Create mask from basin boundary
            basin_mask = rasterio.features.geometry_mask(
                basin_boundary.geometry,
                out_shape=dtm.shape,
                transform=transform,
                invert=True
            )
            # Apply mask to limit processing extent
            basin_mask = basin_mask.astype(bool)
        else:
            basin_mask = np.ones_like(dtm, dtype=bool)
        
        # Reproject river lines to match DTM CRS if needed
        if river_lines.crs != crs:
            river_lines = river_lines.to_crs(crs)
        
        # Rasterize river lines
        river_mask = rasterio.features.geometry_mask(
            river_lines.geometry,
            out_shape=dtm.shape,
            transform=transform,
            invert=True
        )
        
        # Subtract river depth from DTM where rivers exist
        modified_dtm = dtm.copy()
        modified_dtm[river_mask] -= river_depth
        
        # If basin boundary was provided, only update within that area
        if basin_boundary is not None:
            modified_dtm = np.where(basin_mask, modified_dtm, dtm)
        
        # Save the modified DTM if output path is provided
        if output_path:
            with rasterio.open(output_path, 'w', **profile) as dst:
                dst.write(modified_dtm, 1)
            print(f"Modified DTM saved to {output_path}")
            
        return modified_dtm, transform, crs

```

```python
def burn_rivers_into_dtm(dtm_path, river_lines, river_depth, output_path=None, basin_boundary=None):
    """Burns rivers into a DTM with adjacent pixels at half depth."""
    with rasterio.open(dtm_path) as src:
        dtm = src.read(1)
        transform = src.transform
        crs = src.crs
        profile = src.profile.copy()
        
        # Handle basin boundary
        if basin_boundary is not None:
            if basin_boundary.crs != crs:
                basin_boundary = basin_boundary.to_crs(crs)
            basin_mask = rasterio.features.geometry_mask(
                basin_boundary.geometry,
                out_shape=dtm.shape,
                transform=transform,
                invert=True
            )
        else:
            basin_mask = np.ones_like(dtm, dtype=bool)
        
        # Handle river lines
        if river_lines.crs != crs:
            river_lines = river_lines.to_crs(crs)
        
        river_mask = rasterio.features.geometry_mask(
            river_lines.geometry,
            out_shape=dtm.shape,
            transform=transform,
            invert=True
        )
        
        # Create dilated mask for adjacent pixels
        from scipy import ndimage
        dilated_river_mask = ndimage.binary_dilation(river_mask)
        adjacent_pixels = dilated_river_mask & ~river_mask
        
        # Modify DTM
        modified_dtm = dtm.copy()
        modified_dtm[river_mask] -= river_depth  # Full depth at river pixels
        modified_dtm[adjacent_pixels] -= river_depth / 2  # Half depth at adjacent pixels
        
        # Apply basin mask if provided
        if basin_boundary is not None:
            modified_dtm = np.where(basin_mask, modified_dtm, dtm)
        
        # Save output
        if output_path:
            with rasterio.open(output_path, 'w', **profile) as dst:
                dst.write(modified_dtm, 1)
        
        return modified_dtm, transform, crs

```
